```python
import os
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
```

```python
matlab_data = np.loadtxt("../data/raw/matlab_uniform_5000.csv", delimiter=',', skiprows=1)
python_data = np.loadtxt("../data/raw/python_matlab_uniform_5000.csv", delimiter=',', skiprows=1)
```

```python
KE_diffs = matlab_data[:, 8] - python_data[:, 8] 
```

```python
plt.hist(KE_diffs, bins=np.arange(-1e-3, 1e-3, 1e-5));
plt.title("Distribution of KE ratio differences between Matlab and Python code is small");
plt.ylabel("Frequency");
plt.xlabel("Difference");
```
